```
#1

# Estado inicial
saldo = 500_000      # Saldo de la cuenta
deposito = 150_000   # Monto a depositar
retiro = 80_000      # Monto a retirar

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero

interes = 0.02
saldo_final = saldo + (saldo * interes)  #saldo_final como el saldo actualizado con el interés aplicado.

# Salida del estado final
print("Saldo final de la cuenta:", saldo)
print(f"Saldo final con interés: ${saldo_final:.2f}")




#Version con entrada de usuario:
# Solicitar datos al usuario
saldo_inicial = float(input("Ingrese el saldo inicial de la cuenta: "))
deposito = float(input("Ingrese el monto a depositar: "))
retiro = float(input("Ingrese el monto a retirar: "))
interes = float(input("Ingrese la tasa de interés mensual (ej: 0.02 para 2%): "))

# Estado inicial
saldo = saldo_inicial

# Actualización del estado
saldo = saldo + deposito
saldo = saldo - retiro

# Aplicar interés
saldo_final = saldo + (saldo * interes)


print(f"Saldo después de depósitos y retiros: ${saldo:.2f}")
print(f"Saldo final con interés: ${saldo_final:.2f}")


"""¿Qué representa el estado inicial y el estado final del programa?
Estado inicial: Es la situación del programa antes de realizar cualquier operación(saldo original de la cuenta bancaria
antes de hacer depósitos, retiros o aplicar intereses)

Estado final: Es la situación del programa después de ejecutar todas las transformaciones"""

""" Cómo cada instrucción afecta la memoria.
saldo = 500_000	Se crea una variable saldo y se guarda el valor 500000 en memoria
deposito = 150_000	Se crea deposito con valor 150000
retiro = 80_000	Se crea retiro con valor 80000
saldo = saldo + deposito	lee saldo (500000) y deposito (150000), suma ambos (650000), y sobrescribe saldo con el nuevo valor
saldo = saldo - retiro	Lee saldo (650000) y retiro (80000), resta (570000), y sobrescribe nuevamente saldo
interes = 0.02	Crea interes con valor 0.02
saldo_final = saldo + (saldo * interes)	Lee saldo (570000) e interes (0.02), calcula (570000 + 11400 = 581400), y crea saldo_final con ese valor
print(...)	Lee las variables necesarias y las muestra en pantalla (sin modificar la memoria)
"""


#2

# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

# Operaciones basicas
print("Temperaturas originales:", temperaturas)


# Agregar temperatura al final
temperaturas.append(23.7)
print("Después de agregar 23.7:", temperaturas)

# Eliminar primera temperatura
temperaturas.pop(0)
print("Después de eliminar la primera:", temperaturas)

# Insertar temperatura en segunda posición
temperaturas.insert(1, 24.0)
print("Después de insertar 24.0:", temperaturas)

# 4. Mostrar lista ordenada
temperaturas_ordenadas = sorted(temperaturas)
print("Lista ordenada:", temperaturas_ordenadas)

# 5. Calcular diferencia y promedio
max_temp = max(temperaturas)
min_temp = min(temperaturas)
diferencia = max_temp - min_temp
promedio = sum(temperaturas) / len(temperaturas)

print(f"Diferencia max-min: {diferencia:.2f}°C")
print(f"Promedio: {promedio:.2f}°C")

# 6. Encontrar índice de 23.7
indice = temperaturas.index(23.7) if 23.7 in temperaturas else -1
print(f"Índice de 23.7: {indice}")

# 7. Temperaturas de posición 2 a 4
print("Posiciones 2 a 4:", temperaturas[1:4])

# 8. Tres primeras temperaturas
print("Tres primeras:", temperaturas[:3])


#3

# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

# Mostrar nombre del primer producto
print("Primer producto:", inventario[0]["nombre"])

# Agregar nuevo producto
inventario.append({"nombre": "Azúcar", "precio": 2000, "stock": 25})
print("Inventario después:", inventario)

# Disminuir stock de Leche en 2
for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
        break

# 4. Calcular valor total del inventario
valor_total = sum(p["precio"] * p["stock"] for p in inventario)
print(f"Valor total del inventario: ${valor_total}")

# 5. Mostrar productos con stock < 10
print("Productos con stock menor a 10:")
for producto in inventario:
    if producto["stock"] < 10:
        print(f"  {producto['nombre']}: stock {producto['stock']}")

# 6 y 7. Aumentar precio en 10% y redondear
for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)

print("Precios después del aumento del 10%:")
for producto in inventario:
    print(f"  {producto['nombre']}: ${producto['precio']}")

# 8. Buscar producto por nombre (usuario)
buscar = input("Ingrese nombre del producto a buscar: ")
encontrado = None
for producto in inventario:
    if producto["nombre"].lower() == buscar.lower():
        encontrado = producto
        break

if encontrado:
    print(f"Producto encontrado: {encontrado}")
else:
    print("Producto no encontrado")

# 9. Eliminar producto si stock es 0
inventario = [p for p in inventario if p["stock"] > 0]
print("Inventario después de eliminar stock=0:", inventario)

# 10. Ordenar lista por precio ascendente
inventario.sort(key=lambda x: x["precio"])
print("Inventario ordenado por precio:")
for producto in inventario:
    print(f"  {producto['nombre']}: ${producto['precio']}")




#4

# Tupla: coordenadas (x, y)
punto = (3, 7)
print("Punto:", punto)

# 1. Crear tupla rectangulo
rectangulo = (10, 5)

# 2. Calcular área y perímetro
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])
print(f"Área: {area}, Perímetro: {perimetro}")

# 3. Intentar cambiar el valor de ancho
rectangulo[0] = 15
print("Se obtiene type error porque Las tuplas son inmutables, no se pueden modificar después de creadas.")

# 4. Calcular distancia total recorrida
posiciones = [(0, 0), (1, 2), (3, 4)]

distancia_total = 0
for i in range(len(posiciones) - 1):
    x1, y1 = posiciones[i]
    x2, y2 = posiciones[i + 1]
    distancia = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
    distancia_total += distancia

print(f"Distancia total recorrida: {distancia_total:.2f}")


#5
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

# 1. Mostrar cuántos libros prestados y disponibles
prestados = sum(1 for libro in biblioteca["libros"] if libro["prestado"])
disponibles = len(biblioteca["libros"]) - prestados
print(f"Libros prestados: {prestados}")
print(f"Libros disponibles: {disponibles}")

# 2. Agregar un nuevo libro
biblioteca["libros"].append({
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "prestado": False
})

# 3. Marcar como prestado el libro "1984"
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        break

# 4. Mostrar estado actualizado
print("\nEstado actualizado de la biblioteca:")
print(biblioteca)


]
```
