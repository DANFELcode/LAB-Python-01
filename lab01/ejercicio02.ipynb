```
# EJERCICIO 2 - INSTRUCCIONES SIMPLES Y ESTRUCTURADAS

#1
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ")

#Nuevo requisito
sancionado = input("¿Está sancionado? (s/n): ")

if edad >= 18 and tiene_licencia == "s" and sancionado == "n":
    print("Puede conducir")
else:
    print("No puede conducir")
    if edad < 18:
        print("Edad insuficiente (mínimo 18 años)")
    if tiene_licencia != "s":
        print("No tiene licencia de conducir")
    if sancionado == "s":
        print("Está sancionado")


#2
for i in range(1, 6):
    print(f"Iteración número {i}")

# Suma de los primeros 5 números naturales
suma = 0
for i in range(1, 6):
    suma += i
print(f"Suma: {suma}")

# Imprimir solo números pares
for i in range(1, 6):
    if i % 2 == 0:
        print(f"Par: {i}")

# Acumulador y resultado final
acumulador = 0
for i in range(1, 6):
    acumulador += i
print(f"Resultado: {acumulador}")

# Usando el ejercicio anterior de temperaturas, encontrar las mayores al promedio
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
promedio = sum(temperaturas) / len(temperaturas)
print(f"Promedio: {promedio}")
for temp in temperaturas:
    if temp > promedio:
        print(f"Mayor al promedio: {temp}")


#4
suma_pares = 0
suma_impares = 0  # Nueva variable para acumular impares

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

print("Suma de los pares:", suma_pares)
print("Suma de los impares:", suma_impares)

# Estado final como diccionario
estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}
print("Estado final:", estado_final)

# Explicación del estado dentro del bucle
#El estado cambia en cada iteración:
for i in range(1, 11):
    if i % 2 == 0:
        print(f"  i={i}: suma_pares → {suma_pares} + {i} = {suma_pares + i}")
    else:
        print(f"  i={i}: suma_impares → {suma_impares} + {i} = {suma_impares + i}")
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i
print(f"\nEstado final después del bucle:")
print(f"  suma_pares = {suma_pares}, suma_impares = {suma_impares}")




#5
contador = 0
while contador < 5:
    print("Contador:", contador)
    contador += 1

# Modificado para contar hacia atrás
contador = 5
while contador > 0:
    print("Contador:", contador)
    contador -= 1

# Programa que solicita números hasta que ingrese 0
numero = int(input("Ingrese un número: "))
while numero != 0:
    numero = int(input("Ingrese un número: "))
print("Finalizó")

# Explicación del estado del ciclo
# El estado del ciclo es la variable 'contador' porque su valor determina si la condición se cumple o no.


productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    # Usar continue para saltar productos con precio negativo o cantidad menor a 0
    if precio < 0 or cantidad < 0:
        print("Precio o cantidad inválidos. Producto no agregado.")
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)

# Mostrar el producto más caro usando max()
if productos:
    producto_mas_caro = max(productos, key=lambda x: x["precio"])
    print(f"Producto más caro: {producto_mas_caro['nombre']} - ${producto_mas_caro['precio']}")

# Estado final del inventario como lista de diccionarios
print("\nEstado final del inventario:")
for p in productos:
    print(p)
```
